```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
filename = "in_class_test(in).csv"
df = pd.read_csv(filename)
```

```python

df.head()
```

i need to check and clean the data

```python
# Check first few rows
display(df.head())

# Check data types and missing values
print(df.info())

# Statistical summary of numerical columns
display(df.describe())
```

the sparse columns suggest that some data is not collected annually for every country, requiring significant cleaning thatr we are going to do now + see if there are missing values

```python
# 1. Check for missing values
missing_percent = df.isnull().mean() * 100
print(missing_percent[missing_percent > 0].sort_values(ascending=False))

# 2. Drop columns with too much missing data (e.g., > 50% missing)
cols_to_drop = missing_percent[missing_percent > 50].index
df_clean = df.drop(columns=cols_to_drop)

# 3. Drop rows where the Target (Life Expectancy) is missing
# "No labels = no (supervised) learning" [cite: 785]
df_clean = df_clean.dropna(subset=['Life Expectancy'])

print(f"Original shape: {df.shape}, Cleaned shape: {df_clean.shape}")
```

I observed that the dataset contains a significant number of missing values + some rows did not have a value

I decided to drop the columns with excessive missing data because imputing guessing values for over 90% of the dataset would not be accurate enough

now that the data are more clean, i will analize the distribution of life expectancy

```python
# Visualizing the distribution of Life Expectancy
plt.figure(figsize=(10, 6))
sns.histplot(df_clean['Life Expectancy'], kde=True, bins=30)
plt.title('Distribution of Life Expectancy')
plt.xlabel('Age (Years)')
plt.show()
```

we cann see that the histogram shows a distribution centered around 60 to 80 years, but with a "tail" extending down to 30-40 years.
this is a a left-skewed distribution.
Most data points indicate a relatively high life expectancy (over 60), but the long tail to the left reveals significant global inequality.
The extremely low values (e.g., ~25 years) might correspond to countries facing severe conflicts or health crises during specific time periods.

now we will do Feature Engineering to put the data into numbers

```python
# One-Hot Encoding for Gender
# We use 'drop_first=True' to avoid redundancy (dummy variable trap)
df_encoded = pd.get_dummies(df_clean, columns=['Gender'], drop_first=True)

# Display result
display(df_encoded[['Country', 'Year', 'Gender_Female', 'Gender_Male']].head())
```

now we will look  for features that strongly predict the label.

```python
# Correlation Matrix to find top predictors
plt.figure(figsize=(12, 8))
# Select only numeric columns
numeric_df = df_encoded.select_dtypes(include=[np.number])
# Get correlation with Life Expectancy
correlations = numeric_df.corr()['Life Expectancy'].sort_values(ascending=False).head(10)
print(correlations)

# Plotting the strongest predictor (e.g., Universal Health Care)
if 'Universal Heath Care Coverage' in df_encoded.columns:
    sns.scatterplot(data=df_encoded, x='Universal Heath Care Coverage', y='Life Expectancy', alpha=0.5)
    plt.title('Life Expectancy vs Health Care Coverage')
    plt.show()
```

we can see that features like Universal Heath Care Coverage and Basic Sanitation Services have a very high positive correlation with Life Expectancy.

Access to healthcare and sanitation are the strongest predictors of a long life in this dataset. These features will likely carry the most weight in any predictive model.
